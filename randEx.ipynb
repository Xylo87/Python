```python
import string

def getRandomString(length):
    letters = string.ascii_letters
    strResult = ''.join(random.choice(letters) for i in range(length))

    return strResult

getRandomString(5)
```

```python
def RandPass(length):
    try:
        if length < 10:
            raise Exception('Password length must be at least 10 characters long')
            
        randomSource = string.ascii_letters + string.digits + string.punctuation

        password = random.sample(string.ascii_uppercase, 2)
        password += random.choice(string.digits)
        password += random.choice(string.punctuation)
        password += random.sample(randomSource, length - 4)

        passToList = list(password)
        random.shuffle(passToList)

        password = ''.join(passToList)

        return password

    except Exception as valueError:
        print(valueError)

finalPass = RandPass(10)
print(finalPass)
```

```python
from datetime import datetime

def getRandomDate(startDate, endDate):
    dateFormat = '%d/%m/%Y'

    ToDateStart = datetime.strptime(startDate, dateFormat)
    ToDateEnd = datetime.strptime(endDate, dateFormat)

    startTimestamp = int(ToDateStart.timestamp())
    endTimestamp = int(ToDateEnd.timestamp())

    randomTS = random.randrange(startTimestamp, endTimestamp)

    randomDate = datetime.fromtimestamp(randomTS)

    return datetime.strftime(randomDate, '%d %B %Y')
    
getRandomDate("1/1/2016", "12/12/2018")
```
