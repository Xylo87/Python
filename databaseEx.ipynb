```python
import sqlite3

try:
    sqliteConnection = sqlite3.connect('python_db.sqlite')
    cursor = sqliteConnection.cursor()
    print('Database created and successfully connected to SQLite')

    sqliteSelectQuery = 'select sqlite_version()'
    cursor.execute(sqliteSelectQuery)
    record = cursor.fetchone()
    print('SQLite Database Version is', record)
    cursor.close()

except sqlite3.Error as error:
    print('Error while connecting to SQLite', error)

finally:
    if sqliteConnection:
        sqliteConnection.close()
        print('The SQLite connection is closed')
```

```python
def get_hospital_detail(hospital_id):

    try:
        sqliteConnection = sqlite3.connect('python_db.sqlite')
        cursor = sqliteConnection.cursor()
        # print('Database created and successfully connected to SQLite')

        sqliteSelectQuery = """SELECT *
        FROM Hospital
        WHERE Hospital_Id = :id"""

        cursor.execute(sqliteSelectQuery, {'id' : hospital_id})
        record = cursor.fetchall()
        # print(record)

        if len(record) > 0:
            print('Printing Hospital record')
            print(f'Hospital ID : {record[0][0]}')
            print(f'Hospital Name : {record[0][1]}')
            print(f'Bed Count : {record[0][2]}')
            print('\n')
        
        else:
            raise Exception('ID you\'re looking for does not exist')

        cursor.close()

    except sqlite3.Error as error:
        print('Error while connecting to SQLite', error)

    except Exception as idError:
        print(idError)

    finally:
        if sqliteConnection:
            sqliteConnection.close()
            # print('The SQLite connection is closed')


get_hospital_detail(2)



def get_doctor_detail(doctor_id):

    try:
        sqliteConnection = sqlite3.connect('python_db.sqlite')
        cursor = sqliteConnection.cursor()
        # print('Database created and successfully connected to SQLite')

        sqliteSelectQuery = """SELECT *
        FROM Doctor
        WHERE Doctor_Id = :id"""

        cursor.execute(sqliteSelectQuery, {'id' : doctor_id})
        record = cursor.fetchall()
        # print(record)

        if len(record) > 0:
            print('Printing Doctor record')
            print(f'Doctor ID : {record[0][0]}')
            print(f'Doctor Name : {record[0][1]}')
            print(f'Hospital ID : {record[0][2]}')
            print(f'Joining Date : {record[0][3]}')
            print(f'Speciality : {record[0][4]}')
            print(f'Salary : {record[0][5]}')
            print(f'Experience : {record[0][6]}')
            print('\n')
        
        else:
            raise Exception('ID you\'re looking for does not exist')

        cursor.close()

    except sqlite3.Error as error:
        print('Error while connecting to SQLite', error)

    except Exception as idError:
        print(idError)

    finally:
        if sqliteConnection:
            sqliteConnection.close()
            # print('The SQLite connection is closed')


get_doctor_detail(104)
```
