```python
import sqlite3

try:
    sqliteConnection = sqlite3.connect('python_db.sqlite')
    cursor = sqliteConnection.cursor()
    print('Database created and successfully connected to SQLite')

    sqliteSelectQuery = 'select sqlite_version()'
    cursor.execute(sqliteSelectQuery)
    record = cursor.fetchone()
    print('SQLite Database Version is', record)
    cursor.close()

except sqlite3.Error as error:
    print('Error while connecting to SQLite', error)

finally:
    if sqliteConnection:
        sqliteConnection.close()
        print('The SQLite connection is closed')
```

```python
def get_hospital_detail(hospital_id):

    try:
        sqliteConnection = sqlite3.connect('python_db.sqlite')
        cursor = sqliteConnection.cursor()
        # print('Database created and successfully connected to SQLite')

        sqliteSelectQuery = """SELECT *
        FROM Hospital
        WHERE Hospital_Id = :id"""

        cursor.execute(sqliteSelectQuery, {'id' : hospital_id})
        record = cursor.fetchall()
        # print(record)

        if type(hospital_id) is int and len(record) > 0:
            print('Printing Hospital record')
            print(f'Hospital ID : {record[0][0]}')
            print(f'Hospital Name : {record[0][1]}')
            print(f'Bed Count : {record[0][2]}')
            print('\n')
        
        else:
            raise Exception('ID you\'re looking for does not exist')

        cursor.close()

    except sqlite3.Error as error:
        print('Error while connecting to SQLite', error)

    except Exception as idError:
        print(idError)

    finally:
        if sqliteConnection:
            sqliteConnection.close()
            # print('The SQLite connection is closed')


get_hospital_detail(1)



def get_doctor_detail(doctor_id):

    try:
        sqliteConnection = sqlite3.connect('python_db.sqlite')
        cursor = sqliteConnection.cursor()
        # print('Database created and successfully connected to SQLite')

        sqliteSelectQuery = """SELECT *
        FROM Doctor
        WHERE Doctor_Id = :id"""

        cursor.execute(sqliteSelectQuery, {'id' : doctor_id})
        record = cursor.fetchall()
        # print(record)

        if type(doctor_id) is int and len(record) > 0:
            print('Printing Doctor record')
            print(f'Doctor ID : {record[0][0]}')
            print(f'Doctor Name : {record[0][1]}')
            print(f'Hospital ID : {record[0][2]}')
            print(f'Joining Date : {record[0][3]}')
            print(f'Speciality : {record[0][4]}')
            print(f'Salary : {record[0][5]}')
            print(f'Experience : {record[0][6]}')
            print('\n')
        
        else:
            raise Exception('ID you\'re looking for does not exist')

        cursor.close()

    except sqlite3.Error as error:
        print('Error while connecting to SQLite', error)

    except Exception as idError:
        print(idError)

    finally:
        if sqliteConnection:
            sqliteConnection.close()
            # print('The SQLite connection is closed')


get_doctor_detail(101)
```

```python
def get_specialist_doctors_list(speciality, salary):

    try:
        sqliteConnection = sqlite3.connect('python_db.sqlite')
        cursor = sqliteConnection.cursor()
        # print('Database created and successfully connected to SQLite')

        sqliteSelectQuery = """SELECT *
        FROM Doctor
        WHERE Speciality = :speciality AND Salary > :salary"""

        cursor.execute(sqliteSelectQuery, {'speciality' : speciality, 'salary' : salary})
        record = cursor.fetchall()
        # print(record)

        if type(speciality) is str and type(salary) is int and len(record) > 0:
            for doctor in record:
                print(f'Doctor ID: {doctor[0]}')
                print(f'Doctor Name: {doctor[1]}')
                print(f'Hospital ID: {doctor[2]}')
                print(f'Joining Date: {doctor[3]}')
                print(f'Speciality: {doctor[4]}')
                print(f'Salary: {doctor[5]}')
                print(f'Experience: {doctor[6]}')
                print('\n')

        else:
            raise Exception('No data found')

        cursor.close()

    except sqlite3.Error as error:
        print('Error while connecting to SQLite', error)

    except Exception as idError:
        print(idError)

    finally:
        if sqliteConnection:
            sqliteConnection.close()
            # print('The SQLite connection is closed')


get_specialist_doctors_list('Garnacologist', 30000)
```

```python
def get_doctors(hospital_id):

    try:
        sqliteConnection = sqlite3.connect('python_db.sqlite')
        cursor = sqliteConnection.cursor()
        # print('Database created and successfully connected to SQLite')

        sqliteSelectQuery = """SELECT D.*, H.Hospital_Name
        FROM Doctor AS D
        INNER JOIN Hospital AS H ON H.Hospital_Id = D.Hospital_ID
        WHERE D.Hospital_Id = :id"""

        cursor.execute(sqliteSelectQuery, {'id' : hospital_id})
        record = cursor.fetchall()
        # print(record)

        if type(hospital_id) is int and len(record) > 0:
            for doctor in record:
                print(f'Doctor ID: {doctor[0]}')
                print(f'Doctor Name: {doctor[1]}')
                print(f'Hospital ID: {doctor[2]}')
                print(f'Hospital Name: {doctor[7]}')
                print(f'Joining Date: {doctor[3]}')
                print(f'Speciality: {doctor[4]}')
                print(f'Salary: {doctor[5]}')
                print(f'Experience: {doctor[6]}')
                print('\n')

        else:
            raise Exception('No data found')

        cursor.close()

    except sqlite3.Error as error:
        print('Error while connecting to SQLite', error)

    except Exception as idError:
        print(idError)

    finally:
        if sqliteConnection:
            sqliteConnection.close()
            # print('The SQLite connection is closed')


get_doctors(3)
```

```python
from datetime import datetime
from dateutil.relativedelta import relativedelta

def update_doctor_experience(doctor_id):

    try:
        sqliteConnection = sqlite3.connect('python_db.sqlite')
        cursor = sqliteConnection.cursor()
        # print('Database created and successfully connected to SQLite')


        sqliteSelectQuery = """SELECT Joining_Date
        FROM Doctor
        WHERE Doctor_Id = :id"""

        cursor.execute(sqliteSelectQuery, {'id' : doctor_id})
        joiningDate = cursor.fetchone()
        
        if len(joiningDate) > 0:
            today = datetime.now().date()
            dateFormat = '%Y-%m-%d'

            JDtoDate = datetime.strptime(joiningDate[0], dateFormat).date()
            expInYears = relativedelta(today, JDtoDate).years

        else:
            raise Exception('No data found')

        
        sqliteSelectQuery = """UPDATE Doctor
        SET Experience = :exp
        WHERE Doctor_Id = :id"""

        cursor.execute(sqliteSelectQuery, {'exp' : expInYears, 'id' : doctor_id})
        sqliteConnection.commit()


        cursor.close()

    except sqlite3.Error as error:
        print('Error while connecting to SQLite', error)

    except Exception as idError:
        print(idError)

    finally:
        if sqliteConnection:
            sqliteConnection.close()
            # print('The SQLite connection is closed')


update_doctor_experience(101)
```
