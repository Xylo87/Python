```python
from json import JSONEncoder

class Vehicle:
    def __init__(self, name, engine, price):
        self.name = name
        self.engine = engine
        self.price = price

    # def convertJSON(self):
    #     toDict = {'name': self.name, 'engine': self.engine, 'price': self.price}
    #     return json.dumps(toDict, indent=4, sort_keys=True)

    def convertJSON(self):
        return json.dumps(self.__dict__, indent=4, sort_keys=True)
    
# class VehicleEncoder(JSONEncoder):
#         def default(self, o):
#             return o.__dict__

vehicle = Vehicle("Toyota Rav4", "2.5L", 32000)

print(vehicle.convertJSON())

# encoderClass = json.dumps(vehicle, indent=4, cls=VehicleEncoder)
# print(encoderClass)


# print(vars(vehicle))
# print([m for m in dir(vehicle) if callable(getattr(vehicle, m)) and not m.startswith('_')])
```

```python
def vehicleDecoder(obj):
    return Vehicle(obj['name'], obj['engine'], obj['price'])

vehObj = {"name": "Toyota Rav4", "engine": "2.5L", "price": 32000 }
vehObjJSON = json.dumps(vehObj)

vehObjToObj = json.loads(vehObjJSON, object_hook=vehicleDecoder)
print(vehObjToObj.name, vehObjToObj.engine)
```

```python
def validateJSON(JSONdata):
    try:
        json.loads(JSONdata)
    except ValueError as e:
        return False
    return True

InvalidJsonData = """{ "company":{ "employee":{ "name":"emma", "payble":{ "salary":7000 "bonus":800} } } }"""
isValid = validateJSON(InvalidJsonData)

print(isValid)
```
